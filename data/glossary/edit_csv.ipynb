```python
# Source - https://stackoverflow.com/a/15741627
# Posted by jamylak, modified by community. See post 'Timeline' for change history
# Retrieved 2026-09-28, License - CC BY-SA 4.0
import pandas as pd

df = pd.read_csv("glossary_deduplicated.csv", sep = ";")
df_no_duplicates_columns = df.drop_duplicates(subset=['Concept'])
df_no_duplicates_columns.to_csv("glossary_bagofwords_deduplicated.csv", index=False, sep = ";", encoding='utf-8-sig')
```

```python
from pathlib import Path

path = Path("glossary_deduplicated.csv")
raw = path.read_bytes()

print("First 100 bytes:")
print(raw[:100])

print("\nUTF-8 test:")
try:
    text = raw.decode("utf-8")
    print("VALID UTF-8")
except UnicodeDecodeError as e:
    print("NOT valid UTF-8")
    print(e)
```
